````python
#LLM solutions
# Solution 1: nested list comprehension
def create_board_v1(n):
    return [[0] * n for _ in range(n)]


# Solution 2: build the rows with a loop
def create_board_v2(n):
    board = []
    for _ in range(n):
        row = []
        for _ in range(n):
            row.append(0)
        board.append(row)
    return board


# Solution 3: use map to create one row for each board row
def create_board_v3(n):
    return list(map(lambda _: [0] * n, range(n)))


# Test all three solutions
display_board(create_board_v1(3))
display_board(create_board_v2(3))
display_board(create_board_v3(3))
````

*Exercise 2:* Write a function that takes two integers `n` and `m` and **draws** an `n` by `m` game board.

For example, the following is a 3×3 board:

```
   --- --- --- 
  |   |   |   | 
   --- --- ---  
  |   |   |   | 
   --- --- ---  
  |   |   |   | 
   --- --- --- 
```


````python
# Write your solution here

class GameBoard:
    def __init__(self, board):
        self.board = board
        self.rows = len(board)
        self.cols = len(board[0])
        self.symbols = {
            0: " ",
            1: "X",
            2: "O",
        }

    def draw_board(self):
        for row in self.board:
            print(" ---" * self.cols)
            print("|", end="")
            for cell in row:
                print(f" {cell} |", end="")
            print()
        print(" ---" * self.cols)
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def check_board(self):
    def is_winning_line(line):
        return line[0] != 0 and len(set(line)) == 1

    for row in self.board:
        if is_winning_line(row):
            return row[0]

    for col in range(self.cols):
        column = [self.board[row][col] for row in range(self.rows)]
        if is_winning_line(column):
            return column[0]

    diagonal = [self.board[i][i] for i in range(self.rows)]
    if is_winning_line(diagonal):
        return diagonal[0]

    diagonal = [self.board[i][self.rows - 1 - i] for i in range(self.rows)]
    if is_winning_line(diagonal):
        return diagonal[0]

    if any(0 in row for row in self.board):
        return -1

    return 0

#attach method to class
GameBoard.check_board = check_board
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def place_marker(self, row, col, player_num):
    if self.board[row][col] != 0:
        print(f"Invalid move to {row}, {col}. Try again.")
        return False

    self.board[row][col] = player_num
    self.draw_board()
    return True

GameBoard.place_marker = place_marker
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
import copy #used in lecture notes to make sure to not replace true board cells


def computer_move(self, computer_num=2):
    opponent_num = 1 if computer_num == 2 else 2

    def empty_positions(board): #returns all empty positions 
        return [
            (row, col)
            for row in range(len(board))
            for col in range(len(board[row]))
            if board[row][col] == 0
        ]

    def winning_move(player_num):
        for row, col in empty_positions(self.board): #goes thorugh all empty locations
            trial_board = copy.deepcopy(self.board)
            trial_board[row][col] = player_num #sets empty locations to player marker to see if win
            trial_game = GameBoard(trial_board) 

            if trial_game.check_board() == player_num: #if winning move is present it returns the move coords else NONE
                return row, col

        return None

    move = winning_move(computer_num) #performs winning move

    if move is None: #if it cant win it blocks the opponent's winning move
        move = winning_move(opponent_num)

    
    if move is None: #if computer has no winning or blocking move it takes center if possible
        center = (self.rows // 2, self.cols // 2)
        if self.board[center[0]][center[1]] == 0:
            move = center

    if move is None: # if none of the above are possible it takes an emptyt corner
        corners = [
            (0, 0),
            (0, self.cols - 1),
            (self.rows - 1, 0),
            (self.rows - 1, self.cols - 1),
        ]
        move = next((position for position in corners if self.board[position[0]][position[1]] == 0), None)

    # if there is no corner no center no winning no blocking move it cries throws a fit and takes the first remaining empty square
    if move is None:
        positions = empty_positions(self.board)
        if not positions:
            return False
        move = positions[0]

    return self.place_marker(move[0], move[1], computer_num)


GameBoard.computer_move = computer_move
````

````python
# Test your solution here

def play_tic_tac_toe_with_computer():
    new_Game_Board = GameBoard(make_empty_board(board_size))

    print("Welcome to Tic Tac Toe!")
    player = 1

    while new_Game_Board.check_board() == -1:

        if player == 1:
            new_Game_Board.make_player_move(player)
        else:
            new_Game_Board.computer_move()

        status = new_Game_Board.check_board()

        if status == 0:
            print("The game is a draw!")

        if player == 1:
            player = 2
        else:
            player = 1

    if status == 1 or status == 2:   
        print(f"Player {status} wins!")

play_tic_tac_toe_with_computer()
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here

def play_computer_game(size, depth):
    game = GameBoard(make_empty_board(size))
    player = 1

    while True:
        if player == 1:
            game.computer_move(1)
        else:
            game.maxdepth_move(depth, 2)

        status = game.check_board()
        if status != -1:
            return status

        player = 2 if player == 1 else 1


def measure_maxdepth_win_rate(size, depth, games=10):
    results = [play_computer_game(size, depth) for _ in range(games)]
    maxdepth_wins = results.count(2)
    return maxdepth_wins / games, results
````

````python
# Test your solution here
for size, depth in [(3, 9), (4, 3), (5, 2)]:
    win_rate, results = measure_maxdepth_win_rate(size, depth)
    print(f"{size}x{size}: max-depth wins {results.count(2)}/10, win rate = {win_rate:.0%}")
````
