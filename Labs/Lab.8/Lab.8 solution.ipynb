### Exercise 3: Training a Classifier

Let's repeat what we did by hand in the previous lab (Lab 7, Exercise 8) using scikit-learn. We'll use an LDA classifier, which we can instantiate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get an unbiased assessment of the performance of machine learning algorithms)

For example, here we'll use 4M events for training and the remainder for testing.

```python
N_Train=4000000
N_light_train = 1500000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]

# training for lighter loads so wsl doesnt crash under resource exhaustion lol
Train_sample_light = df[:N_light_train]
Test_sample_light = df[N_light_train:]
```

We can train the classifier as follows:

```python
Fisher.fit(X_Train,y_Train)
Fisher.fit(X_Test,y_Test)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare the ROC curves computed on the test and on the training samples, in a single plot. Do you see a bias? (Use `Fisher.decision_function` on each sample, and `roc_curve` as in Exercise 5.)

No I do not see a bias, the curves are identical which indicates good model generalization and little overfitting.

```python
fpr, tpr, _ = roc_curve(y_Test, Fisher.decision_function(X_Test))
roc_auc = auc(fpr, tpr)
plt.plot(fpr,tpr,color='darkorange',label='ROC curve (area = %0.2f)' % roc_auc)

fpr_Train, tpr_Train, _ = roc_curve(y_Train, Fisher.decision_function(X_Train))
roc_auc_Train = auc(fpr_Train, tpr_Train)
plt.plot(fpr_Train,tpr_Train,color='blue',label='ROC curve (area = %0.2f)' % roc_auc_Train)

plt.legend(loc="lower right")
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.show()
```

The best method in part B was randomForest. However it took the longest to train so i will instead use linear regression as it had a AUC score of 0.86 compared to RandomForest's AUC = 0.87. Performance close enough to justify waiting less time.

```python
# exercise 4.ab

#instances of classifiers 
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB

classifiers = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Gaussian Naive Bayes": GaussianNB(),
    "Random Forest": RandomForestClassifier(n_estimators=40, random_state=86, n_jobs=-1)
}

# AI was used here to help fetch the classifier function for each chosen classifier. I was unaware hasattr was a function
# so it was helpful to have it suggested.

def get_classifier(classifier, X): 
    if hasattr(classifier, "decision_function"):
        return classifier.decision_function(X)
    elif hasattr(classifier, "predict_proba"):
        return classifier.predict_proba(X)[:, 1] #selects the probability of the positive class for ROC curve
    else:
        return classifier.predict(X)

def compare_roc_curves(classifier, classifier_name):
    feature_sets = {
        "Raw features": RawNames,
        "Derived features": FeatureNames,
        "All features": VarNames[1:]
    }

    plt.figure(figsize=(8, 6))

    for label, features in feature_sets.items():
        X_train = Train_sample_light[features]
        X_test = Test_sample_light[features]
        Y_train = Train_sample_light["signal"]
        Y_test = Test_sample_light["signal"]

        classifier.fit(X_train, Y_train)
        scores = get_classifier(classifier, X_test)

        fpr, tpr, _ = roc_curve(Y_test, scores)
        roc_auc = auc(fpr, tpr)

        plt.plot(fpr, tpr, label=f"{label} (AUC = {roc_auc:.2f})")

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"{classifier_name}: ROC Comparison")
    plt.legend(loc="lower right")
    plt.show()


for name, classifier in classifiers.items():
    compare_roc_curves(classifier, name)

# my wsl was unexpectedly crashing each time i ran this on the full dataset so i created a lighter 
# dataset to run the classifiers on. 
```

```python
# exercise 5.b 


scenarios = [ # (Ns, Nb)
    (10, 100),
    (100, 1000),
    (1000, 10000), 
    (10000, 100000)
]

#train on whole dataset
LR = LogisticRegression(max_iter=1000)
LR.fit(X_Train, y_Train)

test_scores = LR.decision_function(X_Test)
fpr_LR, tpr_LR, thresholds_LR = roc_curve(y_Test, test_scores)
significance_results = {}

print("Logistic regression fitting complete. Calculating significance for different scenarios...")

for Ns, Nb in scenarios:
    Ns_prime = Ns * tpr 
    Nb_prime = Nb * fpr
    denominator = np.sqrt(Ns_prime + Nb_prime)

    significance = np.divide(Ns_prime, denominator, out=np.zeros_like(Ns_prime), where=denominator > 0)
    best_threshold_index = np.argmax(significance)
    best_threshold = thresholds_LR[best_threshold_index]
    y_pred = (test_scores >= best_threshold).astype(int)

    significance_results[(Ns, Nb)] = {
        "best_threshold": best_threshold,
        "max_significance": significance[best_threshold_index],
        "TPR": tpr[best_threshold_index],
        "FPR": fpr[best_threshold_index], 
        "ROC" : roc_curve(y_Test, test_scores),
        "AUC" : roc_auc_score(y_Test, test_scores),
        "Precision" : precision_score(y_Test, y_pred),
        "Recall" : recall_score(y_Test, y_pred),
        "F1 Score" : f1_score(y_Test, y_pred),
        "Accuracy" : accuracy_score(y_Test, y_pred)
    }
```

```python

print("Metrics calculations complete. Displaying results...")
print(f"Performance metrics of logistic regression classifier for different scenarios:\n")

results_df = pd.DataFrame.from_dict(significance_results, orient="index")
results_df
```
