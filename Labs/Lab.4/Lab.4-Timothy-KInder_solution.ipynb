1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class counter: 
    def __init__(self, start=0, max=12): 
        self.count = start 
        self.max = max

    def increment(self): 
        self.count += 1 
        return self.count

    def decrement(self): 
        self.count -= 1 
        return self.count


    def get_count(self): 
        return self.count

    def __str__(self):
        return f"Current count: {self.count}"

    def reset(self):
        self.count = 0
        print("Counter has been reset to 0.")

    def ask_for_reset(self):
        response = input("Would you like to reset the counter? (y/n): ").strip().lower()
        if response == 'y':
            self.reset()
        elif response == 'n':
            print("Counter not reset.")
        else:
            print("Invalid input. Please enter 'y' or 'n'.")
            self.ask_for_reset()

    def interact(self):
        while True:
            print(self)
            action = input("Enter 'i' to increment, 'd' to decrement, 'r' to reset, or 'q' to quit: ").strip().lower()

            if action == 'i':
                if self.count < self.max:
                    self.increment()
                else: 
                    print(f"Counter has reached its maximum value of {self.max}.")
                    self.ask_for_reset()

            elif action == 'd':
                self.decrement()

            elif action == 'r':
                self.reset()

            elif action == 'q':
                print("Exiting counter.")
                break

            else:
                print("Invalid input. Please try again.")

#test code
counter = counter(10)
#print(counter)

counter.interact()
```

10. Copy the `Canvas` class from lecture into a python file creating a `paint` module. Copy your classes from above into the module and implement paint functions. Implement a `CompoundShape` class. Create a simple drawing demonstrating that all of your classes are working.

```python
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.clear_canvas()

    def set_pixel(self, row, col, char="*"):
        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[" "] * self.width for _ in range(self.height)]

    def v_line(self, x, y, width, **kwargs):
        for row in range(x, x + width):
            self.set_pixel(row, y, **kwargs)

    def h_line(self, x, y, height, **kwargs):
        for col in range(y, y + height):
            self.set_pixel(x, col, **kwargs)

    def line(self, x1, y1, x2, y2, **kwargs):
        delta_x = abs(x2 - x1)
        delta_y = abs(y2 - y1)
        step_x = 1 if x1 < x2 else -1
        step_y = 1 if y1 < y2 else -1
        error = delta_x - delta_y

        while True:
            self.set_pixel(y1, x1, **kwargs)
            if x1 == x2 and y1 == y2:
                break
            double_error = 2 * error
            if double_error > -delta_y:
                error -= delta_y
                x1 += step_x
            if double_error < delta_x:
                error += delta_x
                y1 += step_y

    def display(self):
        print("\n".join("".join(row) for row in self.data))


def paint_shape(self, canvas, char="*"):
    points = self.get_xy_pairs(16)
    if len(points) < 2:
        return

    rounded_points = [(round(x), round(y)) for x, y in points]
    for start, end in zip(rounded_points, rounded_points[1:] + rounded_points[:1]):
        canvas.line(start[0], start[1], end[0], end[1], char=char)


Shape.paint = paint_shape


class CompoundShape(Shape):
    def __init__(self, shapes=None):
        super().__init__(shape_type="compound")
        self.__shapes = list(shapes) if shapes else []

    def add(self, shape):
        self.__shapes.append(shape)

    def get_shapes(self):
        return list(self.__shapes)

    def paint(self, canvas, char="*"):
        for shape in self.__shapes:
            shape.paint(canvas, char)


# Demonstrate all shape types and a compound shape.
canvas = Canvas(30, 15)
compound = CompoundShape([rectangle((2, 2), 8, 4), circle((20, 5), 3), triangle((12, 9), 6, 4)])
compound.paint(canvas)
canvas.display()
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class RasterDrawing:
    def __init__(self):
        self.shapes = {}
        self.shape_names = []

    def add_shape(self, shape, name=None):
        if name is None or name == "":
            name = self.assign_name()

        self.shapes[name] = shape
        self.shape_names.append(name)
        return name

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)

    def paint(self, canvas):
        for name in self.shape_names:
            self.shapes[name].paint(canvas)

    def assign_name(self):
        index = 0
        while f"shape_{index}" in self.shapes:
            index += 1
        return f"shape_{index}"


Raster = RasterDrawing()
Raster.update(canvas)
Raster.add_shape(rectangle((2, 2), 2, 2))
Raster.add_shape(circle((15, 4), 3))
Raster.add_shape(triangle((5, 5), 2, 2))

print(Raster.shape_names)
print(Raster.shapes)
Raster.paint(canvas)
canvas.display()
```
